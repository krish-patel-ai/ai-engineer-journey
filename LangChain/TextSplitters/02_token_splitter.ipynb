```python
!pip install tiktoken
from langchain_text_splitters import TokenTextSplitter
from langchain_core.documents import Document

# ── Why token splitting? ──────────────────────────────────
# LLMs measure context in TOKENS not characters
# 1 token ≈ 0.75 words ≈ 4 characters (rough estimate)
# But it varies:
# "AI"           = 1 token
# "Artificial"   = 2 tokens
# "Intelligence" = 3 tokens
# So character count is not accurate for LLM limits

# TokenTextSplitter counts actual tokens
# More accurate but slower than RecursiveCharacterTextSplitter

splitter = TokenTextSplitter(
    chunk_size=100,     # max 100 TOKENS per chunk (not characters)
    chunk_overlap=10    # 10 tokens overlap
)

sample_text = """LangChain is a framework for developing applications powered by large language models.
It provides tools for chaining multiple components together into pipelines.
The framework supports various LLM providers including OpenAI, Anthropic, and Google.
RAG or Retrieval Augmented Generation is one of the most popular patterns in LangChain.
It combines document retrieval with language model generation to answer questions accurately.
LangGraph extends LangChain with support for stateful multi-agent workflows.
Agents can use tools like web search, calculators, and custom APIs to complete tasks.
Memory systems allow conversations to persist context across multiple turns."""

print("═" * 50)
print("TokenTextSplitter")
print("═" * 50)
print(f"Original: {len(sample_text)} characters")
print()

chunks = splitter.split_text(sample_text)

print(f"Total chunks : {len(chunks)}")
print(f"chunk_size   : 100 tokens")
print(f"chunk_overlap: 10 tokens")
print()

for i, chunk in enumerate(chunks):
    print(f"── Chunk {i+1} ({len(chunk)} chars) ──")
    print(chunk)
    print()

# ── Compare with RecursiveCharacterTextSplitter ───────────
from langchain_text_splitters import RecursiveCharacterTextSplitter

char_splitter = RecursiveCharacterTextSplitter(
    chunk_size=400,    # 100 tokens ≈ 400 characters
    chunk_overlap=40
)

char_chunks = char_splitter.split_text(sample_text)

print("═" * 50)
print("Comparison: Token vs Character splitting")
print("═" * 50)
print(f"TokenTextSplitter     : {len(chunks)} chunks")
print(f"CharacterTextSplitter : {len(char_chunks)} chunks")
print()
print("When to use which:")
print("TokenTextSplitter         → when token limit accuracy matters")
print("RecursiveCharacterSplitter → 90% of cases, faster, good enough")
```
