```python
!pip install langchain-experimental
from langchain_experimental.text_splitter import SemanticChunker
from langchain_community.embeddings import HuggingFaceEmbeddings
from langchain_core.documents import Document
```

```python
embeddings=HuggingFaceEmbeddings(model_name="sentence-transformers/all-MiniLM-L6-v2")
```

```python
splitter=SemanticChunker(embeddings=embeddings,breakpoint_threshold_type="percentile",breakpoint_threshold_amount=85)
sample_text= """LangChain is a framework for building LLM applications. 
It provides components for chains, memory, and agents. 
The framework makes it easy to connect different LLM providers.
LangChain supports OpenAI, Anthropic, Google, and many others.

RAG stands for Retrieval Augmented Generation. 
RAG combines document search with LLM generation. 
It helps LLMs answer questions about your specific documents.
RAG reduces hallucinations by grounding answers in real data.

LangGraph is built on top of LangChain for agentic workflows.
It allows you to build stateful multi-agent applications.
Agents can loop, branch, and make decisions based on state.
Human-in-the-loop is a key feature of LangGraph applications."""
print(f"Original: {len(sample_text)} characters")
```

```python
chunks=splitter.split_text(sample_text)
print(f"Total chunks created: {len(chunks)}")
```

```python
for i, chunk in enumerate(chunks):
    print(f"── Chunk {i+1} ({len(chunk)} chars) ──")
    print(chunk)
    print()
```
