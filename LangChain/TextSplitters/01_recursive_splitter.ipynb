```python
from langchain_text_splitters import RecursiveCharacterTextSplitter
from langchain_community.document_loaders import PyPDFLoader
splitter=RecursiveCharacterTextSplitter(
    chunk_size=200,
    chunk_overlap=50,
    length_function=len,
    add_start_index=True)
```

```python
sample_text="""LangChain is a framework for developing applications powered by language models.

It enables applications that are context-aware and reason about their environment.

The main components of LangChain are chains, agents, and memory.

Chains allow you to combine multiple components together to create a pipeline.

Agents use language models to decide which actions to take and in what order.

Memory allows chains and agents to keep information between calls.

RAG stands for Retrieval Augmented Generation. It combines document retrieval with LLM generation.

Vector databases store embeddings and allow similarity search at scale.

LangGraph is a library for building stateful multi-agent applications."""
print(f"Original text  : {len(sample_text)} characters")
```

```python
chunks=splitter.split_text(sample_text)
print(f"Total chunks   : {len(chunks)}")
```

```python
for i ,chunk in enumerate(chunks):
    print(f"── Chunk {i+1} ({len(chunk)} chars) ──")
    print(chunk)
    print()
```

```python
from langchain_core.documents import Document

docs = [
    Document(
        page_content=sample_text,
        metadata={"source": "langchain_guide.pdf", "page": 1}
    )
]

# split_documents → takes List[Document] → returns List[Document]
# IMPORTANT: metadata is PRESERVED in every chunk
split_docs = splitter.split_documents(docs)

print(f"Original docs  : {len(docs)}")
print(f"After splitting: {len(split_docs)} chunks")
print()

for i, doc in enumerate(split_docs):
    print(f"── Chunk {i+1} ──")
    print(f"Content  : {doc.page_content[:80]}...")
    print(f"Metadata : {doc.metadata}")
    # Notice metadata still has source and page — preserved from original
    print()
```
