```python
!pip install pytube
!pip install youtube-transcript-api
from langchain_community.document_loaders import YoutubeLoader
from langchain_community.document_loaders.youtube import TranscriptFormat
```

```python
video_url="https://www.youtube.com/watch?v=LbT1yp6quS8"
loader=YoutubeLoader.from_youtube_url(video_url,add_video_info=False,language=["en", "en-US", "en-GB"])
docs=loader.load()
print(f"Documents loaded : {len(docs)}")
doc=docs[0]
for key,val in doc.metadata.items():
    print(f"{key}: {val}")
print()
```

```python
print(doc.page_content[:500])
print()
print(f"Total transcript length: {len(doc.page_content)} characters")
print()
```

```python
# For long videos you want chunks with timestamps
loader_chunked = YoutubeLoader.from_youtube_url(
    video_url,
    add_video_info=False,
    language=["en", "en-US", "en-GB"],
    transcript_format=TranscriptFormat.CHUNKS,
    chunk_size_seconds=60,   # 1 chunk per minute
)
chunked_docs = loader_chunked.load()

print("═" * 50)
print("YoutubeLoader — chunked by time")
print("═" * 50)
print(f"Total chunks : {len(chunked_docs)}")
print()
for i, chunk in enumerate(chunked_docs[:3]):
    print(f"Chunk {i+1}: {chunk.metadata}")
    print(f"Text  : {chunk.page_content[:100]}...")
    print()
```
