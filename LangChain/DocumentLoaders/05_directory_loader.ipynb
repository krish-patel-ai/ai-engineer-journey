```python
from langchain_community.document_loaders import DirectoryLoader
from langchain_community.document_loaders import PyPDFLoader
from langchain_community.document_loaders import TextLoader
import os
```

```python
os.makedirs("test_docs", exist_ok=True)
```

```python
for i in range(3):
    with open(f"test_docs/doc{i+1}.txt", "w") as f:
        f.write(f"Document {i+1}\n")
        f.write(f"This is about LangChain topic {i+1}.\n")
        f.write(f"LangChain makes building AI apps easy.\n")
        f.write(f"Document {i+1} content ends here.\n")
```

```python
print("Test folder created with 3 text files")
print()
```

```python
loader = DirectoryLoader(
    "test_docs",              # folder path
    glob="**/*.txt",          # which files to load (* = all .txt)
    loader_cls=TextLoader,    # which loader to use per file
    show_progress=True,       # shows loading progress bar
    use_multithreading=True,  # loads files in parallel — faster
)
docs = loader.load()
```

```python
print(f"Files found and loaded: {len(docs)}")
print()
```

```python
for doc in docs:
    print(f"── Document ──")
    print(f"Source  : {doc.metadata['source']}")
    print(f"Content : {doc.page_content[:100]}")
    print()
```
