```python
from langchain_community.document_loaders import WebBaseLoader
import bs4
```

```python
loader=WebBaseLoader("https://en.wikipedia.org/wiki/LangChain")
docs=loader.load()
print(f"Documents loaded : {len(docs)}")
print(f"Characters       : {len(docs[0].page_content)}")
print(f"Metadata         : {docs[0].metadata}")
print()
print("── First 500 characters ──")
print(docs[0].page_content[:500])
print()
```

```python
# Fix — just use basic loader, filter manually
loader_targeted = WebBaseLoader(
    web_paths=["https://en.wikipedia.org/wiki/LangChain"],
    bs_kwargs={
        "parse_only": bs4.SoupStrainer("p")  # grab all paragraph tags
    }
)
targeted_docs = loader_targeted.load()

print("═" * 50)
print("WebBaseLoader — targeted (paragraphs only)")
print("═" * 50)
print(f"Characters (targeted) : {len(targeted_docs[0].page_content)}")
print(f"Characters (basic)    : 18339")
print()
print("── Targeted first 500 chars ──")
print(targeted_docs[0].page_content[:500])
```
