```python
loader=PyPDFLoader("Get_Started_With_Smallpdf.pdf")
docs=loader.load()
print(f"Total pages loaded : {len(docs)}")
print()
first=docs[0]
print("── First Document ──")
print(f"Type           : {type(first)}")
print(f"page_content   : {first.page_content[:200]}...")
print(f"metadata       : {first.metadata}")
# metadata shows: source filename + page number
print()           
```

# Lazy loader

```python
loader2 = PyPDFLoader("Get_Started_With_Smallpdf.pdf")
for doc in loader2.lazy_load():
    print(f"Loaded page: {doc.metadata['page']} | {len(doc.page_content)} chars")
print()
```
