```python
from dotenv import load_dotenv
from langchain_groq import ChatGroq
from langchain_core.prompts import ChatPromptTemplate
from langchain_core.output_parsers import StrOutputParser

load_dotenv()

# ── Step 1: The LLM (brain) ──────────────────────────────
llm = ChatGroq(model="llama-3.1-8b-instant")
# Free, fast. No credit card needed.
# llama-3.1-8b-instant is the smallest, fastest Llama model on Groq

# ── Step 2: The prompt template (Mad Libs) ───────────────
prompt = ChatPromptTemplate.from_messages([
    ("system", "You are a helpful teacher. Always explain simply."),
    ("human", "Explain {topic} in 3 bullet points.")
])
# "system" = instructions to the LLM about its personality
# "human"  = what the user is asking
# {topic}  = the blank we fill in later

# ── Step 3: The output parser (clean up) ─────────────────
parser = StrOutputParser()
# Converts LLM response object → plain Python string

# ── Step 4: Snap them together ───────────────────────────
chain = prompt | llm | parser
#        ↑         ↑      ↑
#      fill      think   clean

# ── Step 5: Run it ───────────────────────────────────────
result = chain.invoke({"topic": "neural networks"})
# .invoke() = run the chain once
# pass a dict with values for every {blank} in your template

print(result)
```
